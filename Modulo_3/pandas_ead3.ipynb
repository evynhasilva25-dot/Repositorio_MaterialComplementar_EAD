## 1. Criando Dados de Exemplo

Vamos criar alguns dados para praticar:

```python
import pandas as pd
import numpy as np

# Criando dados de alunos
dados_alunos = {
    'Nome': ['Ana Silva', 'João Santos', 'Maria Oliveira', 'Pedro Costa', 'Carla Lima',
             'Roberto Alves', 'Juliana Ferreira', 'Carlos Souza', 'Patrícia Rocha', 'André Pereira'],
    'Idade': [25, 30, 28, 35, 22, 29, 26, 31, 27, 33],
    'Curso': ['Python', 'Java', 'Python', 'JavaScript', 'Python', 
              'Java', 'JavaScript', 'Python', 'Java', 'JavaScript'],
    'Nota': [8.5, 9.0, 7.5, 9.5, 8.0, 8.5, 9.0, 7.0, 9.5, 8.0],
    'Faltas': [2, 0, 3, 1, 2, 1, 0, 4, 1, 2]
}

df_alunos = pd.DataFrame(dados_alunos)
print("Dados dos Alunos:")
print(df_alunos)
```

## 2. Exportando Dados

### Exportando para CSV:

```python
# Exportando para CSV (sem índice)
df_alunos.to_csv('alunos.csv', index=False)
print("Arquivo 'alunos.csv' criado com sucesso!")

# Exportando para CSV com separador diferente
df_alunos.to_csv('alunos_semicolon.csv', sep=';', index=False)
print("Arquivo 'alunos_semicolon.csv' criado com sucesso!")
```

### Exportando para Excel:

```python
# Exportando para Excel
df_alunos.to_excel('alunos.xlsx', index=False, sheet_name='Alunos')
print("Arquivo 'alunos.xlsx' criado com sucesso!")

# Exportando múltiplas planilhas
with pd.ExcelWriter('alunos_multiplas.xlsx') as writer:
    df_alunos.to_excel(writer, sheet_name='Dados_Completos', index=False)
    
    # Criando um resumo por curso
    resumo_curso = df_alunos.groupby('Curso').agg({
        'Nota': ['mean', 'max', 'min'],
        'Faltas': 'mean'
    })
    resumo_curso.to_excel(writer, sheet_name='Resumo_Curso')
    
print("Arquivo 'alunos_multiplas.xlsx' criado com sucesso!")
```

### Exportando para JSON:

```python
# Exportando para JSON
df_alunos.to_json('alunos.json', orient='records', indent=2)
print("Arquivo 'alunos.json' criado com sucesso!")

# Diferentes orientações de JSON
df_alunos.to_json('alunos_split.json', orient='split', indent=2)
print("Arquivo 'alunos_split.json' criado com sucesso!")
```

## 3. Lendo Dados

### Lendo arquivos CSV:

```python
# Lendo CSV criado anteriormente
df_lido_csv = pd.read_csv('alunos.csv')
print("Dados lidos do CSV:")
print(df_lido_csv)

# Lendo CSV com separador diferente
df_lido_csv_sem = pd.read_csv('alunos_semicolon.csv', sep=';')
print("\nDados lidos do CSV com separador ;:")
print(df_lido_csv_sem)
```

### Lendo arquivos Excel:

```python
# Lendo Excel
df_lido_excel = pd.read_excel('alunos.xlsx', sheet_name='Alunos')
print("Dados lidos do Excel:")
print(df_lido_excel)

# Lendo Excel com múltiplas planilhas
arquivo_excel = pd.ExcelFile('alunos_multiplas.xlsx')
df_completo = pd.read_excel(arquivo_excel, sheet_name='Dados_Completos')
df_resumo = pd.read_excel(arquivo_excel, sheet_name='Resumo_Curso')

print("\nResumo por Curso:")
print(df_resumo)
```

### Lendo arquivos JSON:

```python
# Lendo JSON
df_lido_json = pd.read_json('alunos.json')
print("Dados lidos do JSON:")
print(df_lido_json)
```

## 4. Lendo Dados da Web

O Pandas pode ler dados diretamente da web:

```python
# Lendo dados de uma URL (exemplo com dados de exemplo)
url = 'https://raw.githubusercontent.com/datasciencecampus/datasets/master/remote_sensing/2016-05-03_188_road.csv'
try:
    df_url = pd.read_csv(url)
    print("Dados lidos da URL:")
    print(df_url.head())
except:
    print("Erro ao ler URL. Verifique sua conexão com a internet.")
    
# Lendo tabelas HTML da web
try:
    # Exemplo com tabela da Wikipedia
    url_wiki = 'https://pt.wikipedia.org/wiki/Lista_de_estados_do_Brasil_por_popula%C3%A7%C3%A3o'
    tabelas = pd.read_html(url_wiki)
    print(f"\nEncontradas {len(tabelas)} tabelas na página")
    print("Primeira tabela:")
    print(tabelas[0].head())
except:
    print("Erro ao ler tabelas da web.")
```
